# Traditional reconstruction approach
This method is a "historical" method (<a href="https://dl.acm.org/doi/abs/10.1145/133994.134011">link</a>) for reconstructing a surface from a set of points. It consists in taking an oriented point cloud $(x _i , n _i )$, and estimating for any arbitrary point $x$ in the ambient space a signed distance function as : $u(x) = ± min _i ∥x _i - x∥$

The sign is given by the sign of the scalar product $\langle x - x _i, n _i \rangle$.

# Unsigned Distance Function

In this case (paper <a href="https://arxiv.org/pdf/1911.10414">link</a>), the objective is to learn directly on raw point clouds, without pre-processing to predicts normals/orientation of the shape. To reach this objective, the authors notice the following:
- Using the unsigned distance function (absolute value of the predicted SDF) is then necessary
- Carefully choosing the points where to predict the distance is crucial
- Weights initialization is to be changed

The modification to SDF is simple : the loss is now computed by sampling points around each data point $x _i$ , following a centered Gaussian distribution of variance $\sigma²$:

$$
L(\theta) = \sum _i \mathbb{E} _{x \sim N(x _i, \sigma^2)}[(|u _{\theta}(x)| - |\text{SDF} _{\text{GT}}(x)|)^2]
$$

where $\sigma$ is a parameter that you can play with, and $\text{SDF} _{\text{GT}}(x) = \min _j \Vert x - x _j\Vert $ is the unsigned distance from $x$ to the point cloud (e.g., with a KD-tree).

The last linear layer is now initialized too, with weights following $N\left(\frac{\sqrt{\pi}}{\sqrt{n}}, \epsilon\right)$ law ($n=512$ the input size of the last layer, $\epsilon$ tiny, e.g. $10^{-5}$) and bias initialized to -1. The last layer activation is now $\phi(a) = \text{tanh}(a) + \gamma a$. Gamma parameter now equals to 0.5.
